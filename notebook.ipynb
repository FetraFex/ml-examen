```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Chargement des données
df = pd.read_csv('dataset_morpion.csv')
```

Distribution de x_wins et is_draw — le dataset est-il équilibré ?

```python
# 2. Distribution des cibles
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
sns.countplot(x='x_wins', data=df, ax=ax[0], palette='viridis').set_title('Distribution x_wins')
sns.countplot(x='is_draw', data=df, ax=ax[1], palette='magma').set_title('Distribution is_draw')
plt.show()
```

Quelle case est le plus souvent occupée par X dans les états où il gagne ?

```python
# 3. Case préférée de X pour gagner
df_wins = df[df['x_wins'] == 1]
x_cols = [f'c{i}_x' for i in range(9)]
frequence_x = df_wins[x_cols].sum().sort_values(ascending=False)

plt.figure(figsize=(10, 4))
frequence_x.plot(kind='bar', color='skyblue')
plt.title("Cases occupées par X lors d'une victoire")
plt.ylabel("Nombre d'apparitions")
plt.show()
```

Corrélation entre les features et chaque cible — heatmap.


```python
# 4. Heatmap des corrélations
plt.figure(figsize=(10, 8))
corr = df.corr()[['x_wins', 'is_draw']].drop(['x_wins', 'is_draw'])
sns.heatmap(corr, annot=True, cmap='coolwarm', center=0)
plt.title("Corrélation entre les cases et les cibles")
plt.show()
```
